```python
import time
import datetime
import torch

from rl_firm.env import BaselineABMEnv, make_custom_env
from rl_firm.ppo_lstm import train, HyperParameters
```

## 500k no anneal

```python
agent = train(HyperParameters(total_timesteps=500000), 'lstm_500k_no_anneal')
```
