## Some experiment to understand the squashing function

```python
env = make_custom_env()()
obs, info = env.reset()
```

```python
agent = RecurrentPPOAgent(obs_dim=int(np.array(env.observation_space.shape).prod()))
agent.eval()

lstm_state = (torch.zeros(agent.lstm.num_layers, 1, agent.lstm.hidden_size),
              torch.zeros(agent.lstm.num_layers, 1, agent.lstm.hidden_size))
done = torch.zeros(1)
```

```python
hidden, cat_probs, cont_probs, lstm_state = agent._get_probs(torch.Tensor(np.array([obs])), lstm_state, done)
```

```python
hidden.shape
```

```python
u = cont_probs.rsample()
u.shape
```

```python
y = torch.tanh(u)
y
```

```python
a = agent._squash_affine(y)
a
```

```python
cont_probs.log_prob(u)
```

```python
cont_probs.log_prob(u).sum(1)
```

```python
cont_probs.log_prob(u).sum(-1)
```
