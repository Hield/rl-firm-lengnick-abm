```python
import time
import datetime
from pathlib import Path
import numpy as np
import torch

from rl_firm.ppo_lstm_truncated import train, HyperParameters
```

NOTE: The final training that resulted in the agent used in the thesis

- Recurrent PPO with LSTM
- Bounded action in the policy using a tanh squashing function
- Small bonus for survival
- Episode truncated after 500 years

## 2m anneal (NOTE: Rerun)

```python
TIMEOUT_YEARS = 500
agent = train(HyperParameters(total_timesteps=2000000, anneal_lr=True, survival_bonus=0.01, time_limit=TIMEOUT_YEARS*12), 'lstm_truncated_2m_anneal')
save_path = Path("../results/model/lstm_truncated/2m_anneal.pt")
save_path.parent.mkdir(parents=True, exist_ok=True)
torch.save(agent.state_dict(), save_path)
```
